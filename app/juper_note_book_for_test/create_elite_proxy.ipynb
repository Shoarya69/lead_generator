```python
import requests
```

```python
data = requests.get("https://cdn.jsdelivr.net/gh/proxifly/free-proxy-list@main/proxies/protocols/https/data.json")
```

```python
data
```

```python
proxies = data.json()
print(proxies[:2])
```

```python
import pandas as pd
```

```python
valid_proxies = [
    p for p in proxies
    if p["anonymity"] == "elite"
    and p["protocol"] == "http"
]
```

```python
print(valid_proxies)
```

```python
def check_proxy(proxy):
    try:
        r = requests.get(
            "https://api.ipify.org",
            proxies={"http": proxy, "https": proxy},
            timeout=3
        )
        if r.status_code == 200:
            print("ok")
        return r.status_code == 200
    except:
        return False
```

```python
working = []

for p in valid_proxies:
    proxy = p['proxy']
    if check_proxy(proxy):
        working.append({
            "proxy": proxy,
            "country": p['geolocation']['country']
        })
```
