```python
import itertools
```

```python
comb = list(itertools.product(df_2['Country'],df_1['Business']))
print(comb)
```

```python
df_comb = pd.DataFrame(comb,columns=['Country','Business'])
```

```python
df_comb["query"] = df_comb["Business"]+ " in " + df_comb["Country"]
```

```python
df_comb.head()
```

```python
df_comb["Is_use"]=False
df_comb
```

```python
df_tire_1 = pd.read_csv("/home/shoarya/Desktop/leadomator/app/tire_1.csv")
df_tire_2 = pd.read_csv("/home/shoarya/Desktop/leadomator/app/tire_2.csv")
df_tire_3 = pd.read_csv("/home/shoarya/Desktop/leadomator/app/tire_3.csv")
df_tire_3

```

```python
t1_set = set(df_tire_1.iloc[:,0].str.strip())
t2_set = set(df_tire_2.iloc[:,0].str.strip())
t3_set = set(df_tire_3.iloc[:,0].str.strip())
t1_set
```

```python
def get_tier(business):
    b = str(business).strip()
    if b in t1_set:
        return 1
    elif b in t2_set:
        return 2
    elif b in t3_set:
        return 3
    else:
        return None  # ya default 2 rakh sakta hai
    
df_comb["tier"] = 2  # default

df_comb.loc[df_comb["Business"].isin(df_tire_1['TIER 1']), "tier"] = 1
df_comb.loc[df_comb["Business"].isin(df_tire_3['TIRE 3']), "tier"] = 3

```
